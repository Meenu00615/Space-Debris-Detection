```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd


df = pd.DataFrame(data)

# Convert LAUNCH_DATE to datetime
df['LAUNCH_DATE'] = pd.to_datetime(df['LAUNCH_DATE'], format='%d-%m-%Y')

# Set the style for seaborn
sns.set(style="whitegrid")
sns.set_palette("viridis")

# Visualization 1: Launches over the years
plt.figure(figsize=(10, 6))
sns.countplot(x=df['LAUNCH_DATE'].dt.year)
plt.title('Number of Launches Over the Years', fontsize=14)
plt.xlabel('Year', fontsize=7)
plt.ylabel('Number of Launches', fontsize=7)
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for better visibility
plt.show()


# Visualization 3: RCS Size Distribution
plt.figure(figsize=(8, 4))
sns.countplot(x=df['RCS_SIZE'])
plt.title('Distribution of RCS Sizes', fontsize=14)
plt.xlabel('RCS Size', fontsize=14)
plt.ylabel('Count', fontsize=14)
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Sample data with hypothetical launch count over years
launch_data = {
    'Year': np.arange(1958, 1963),
    'LaunchCount': [3, 8, 12, 10, 15]  # Replace with your actual launch count data
}

launch_df = pd.DataFrame(launch_data)

# Reshape the data for regression analysis
X = launch_df['Year'].values.reshape(-1, 1)
y = launch_df['LaunchCount'].values

# Create and fit the linear regression model
model = LinearRegression()
model.fit(X, y)

# Predict future launch trends
future_years = np.arange(1963, 2025).reshape(-1, 1)
future_launch_predictions = model.predict(future_years)

# Plotting historical data and future predictions
plt.figure(figsize=(10, 6))

plt.scatter(X, y, color='blue', label='Historical Data')
plt.plot(future_years, future_launch_predictions, color='red', linestyle='--', label='Regression Prediction')

plt.title('Launch Trends Prediction with Linear Regression')
plt.xlabel('Year')
plt.ylabel('Number of Launches')
plt.legend()
plt.grid(True)
plt.show()
```
